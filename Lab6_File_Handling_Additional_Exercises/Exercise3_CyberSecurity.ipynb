```python
# === EXERCISE 3: CYBERSECURITY INCIDENT REPORT ===
from pathlib import Path
import csv
import json
import shutil

LAST_NAME = "Viana"
STUDENT_ID = "TUPM-26-1953"
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())

DATA_DIR = Path("../data")
OUTPUT_DIR = Path(".")
BACKUP_DIR = Path("../backup")

input_path = DATA_DIR / "Global_Cybersecurity_Threats_2015-2024.csv"
output_csv = OUTPUT_DIR / f"{LAST_NAME}_cyber_incidents.csv"
top5_json = OUTPUT_DIR / f"{LAST_NAME}_cyber_top5.json"
report_txt = OUTPUT_DIR / f"{LAST_NAME}_cyber_report.txt"
backup_json = BACKUP_DIR / top5_json.name

# Rule
TARGET_YEAR = 2015 + SEED_DIGIT

total_records = 0
target_records = []
attack_type_counts = {}
total_loss = 0.0

with open(input_path, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        try:
            if int(row["Year"]) == TARGET_YEAR:
                target_records.append(row)
                attack_type = row["Attack Type"].strip()
                attack_type_counts[attack_type] = attack_type_counts.get(attack_type, 0) + 1
                loss = float(row["Financial Loss (in Million $)"])
                total_loss += loss
        except (ValueError, KeyError):
            continue

avg_loss = total_loss / len(target_records) if target_records else 0

# Top 5 by financial loss
sorted_records = sorted(target_records, key=lambda x: float(x["Financial Loss (in Million $)"]), reverse=True)
top5 = sorted_records[:5]

# Export filtered CSV
with open(output_csv, "w", encoding="utf-8", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=reader.fieldnames)
    writer.writeheader()
    writer.writerows(target_records)

# Export Top 5 JSON
with open(top5_json, "w", encoding="utf-8") as f:
    json.dump(top5, f, indent=4)

# Backup JSON
shutil.copy2(top5_json, backup_json)

# Generate text report
with open(report_txt, "w", encoding="utf-8") as f:
    f.write("CYBERSECURITY INCIDENT REPORT\n")
    f.write("=" * 40 + "\n")
    f.write(f"Target Year: {TARGET_YEAR}\n")
    f.write(f"Total Records in Dataset: [not counted]\n")
    f.write(f"Incidents in Target Year: {len(target_records)}\n\n")
    f.write("Incident Count per Attack Type:\n")
    for atype, count in attack_type_counts.items():
        f.write(f"  - {atype}: {count}\n")
    f.write(f"\nTotal Financial Loss: ${total_loss:.2f} Million\n")
    f.write(f"Average Financial Loss: ${avg_loss:.2f} Million\n\n")
    f.write("Five Costliest Incidents:\n")
    for i, inc in enumerate(top5, 1):
        f.write(f"  {i}. {inc['Attack Type']} — ${inc['Financial Loss (in Million $)']}M\n")
    f.write(f"\nFiltered CSV File: {output_csv.name}\n")
    f.write(f"JSON Summary File: {top5_json.name}\n")
    f.write(f"Backup Verification: {backup_json.exists()}\n")
    f.write(f"Report File: {report_txt.name}\n")

# Output
print("=== EXERCISE 3 RESULTS ===")
print(f"Target Year: {TARGET_YEAR}")
print(f"Total records in dataset: {total_records}")  # ← HERE
print(f"Incidents in target year: {len(target_records)}")
print(f"Incidents in target year: {len(target_records)}")
print("Count per attack type:")
for atype, count in attack_type_counts.items():
    print(f"  {atype}: {count}")
print(f"Total financial loss: ${total_loss:.2f} Million")
print(f"Average financial loss: ${avg_loss:.2f} Million")
print("Five costliest incidents:")
for i, inc in enumerate(top5, 1):
    print(f"  {i}. {inc['Attack Type']} — ${inc['Financial Loss (in Million $)']}M")
print(f"Filtered CSV file: {output_csv.name}")
print(f"JSON summary file: {top5_json.name}")
print(f"Backup verification: {backup_json.exists()}")
print(f"Report file: {report_txt.name}")
```
